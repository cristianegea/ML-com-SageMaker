```python
# Integração do Python com a AWS
import boto3
```

```python
# Criação da variável que vai armazenar o nome do bucket
bucket = "cursoawssagemaker"
s3 = boto3.resource("s3")   # indica que o recurso da AWS que quero buscar é o S3
```

```python
# Download do arquivo 'digit_download.png'
s3.Bucket(bucket).download_file("arquivos/digit.png", "digit_download.png")
```

```python
# Download do arquivo 'house_prices_download.csv'
s3.Bucket(bucket).download_file("arquivos/house_prices.csv", "house_prices_download.csv")
```
