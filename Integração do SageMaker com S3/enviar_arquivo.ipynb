```python
# https://docs.aws.amazon.com/pythonsdk/?id=docs_gateway
import boto3
from sagemaker.s3 import S3Uploader

# boto3 :: biblioteca que permite a integração entre o python e a aws
# S3Uploader :: responsável por enviar os arquivos para S3
```

```python
# Criação de sessão para cria uma conexão com o S3
session = boto3.Session()
```

```python
# Visualização do atributo 'region_name'
session.region_name

# obs.: sa-east-1 = são paulo
```

Criação de um novo bucket no S3

```python
# Criação de um novo bucket dentro do S3
bucket = 'cursoawssagemaker'
session.client('s3').create_bucket(Bucket=bucket,
    CreateBucketConfiguration={"LocationConstraint": session.region_name})
```

Criação de uma pasta

```python
# Criação de uma subpasta
subpasta = "arquivos"
```

```python
# Envio dos arquivos para o jupiter lab
caminho = "s3://{}/{}".format(bucket, subpasta)   # URI da pasta dentro do bucket
print(caminho)

# variável 'bucket' será copiada para o 1º '{}'
# variável 'subpasta' será copiada para o 2º '{}'
```

```python
s3_imagem = S3Uploader.upload("digit.png", caminho)
print(s3_imagem)
```

```python
s3_base = S3Uploader.upload("house_prices.csv", caminho)
print(s3_base)
```
